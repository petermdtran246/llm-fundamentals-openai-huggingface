```python
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()

client = OpenAI()
```

## Generating Text

```python
def generate_text(prompt):
    response = client.chat.completions.create(
        model="gpt-4o-mini",   
        messages=[
            {"role": "user", "content": prompt}
        ],
        max_tokens=10,
        temperature=0.7,
    )
    return response.choices[0].message.content.strip()
```

```python
prompt = "Once upon a time"
```

```python
generated_text = generate_text(prompt)
print(prompt, generated_text)
```

## Customizing the Output

## Summarising Text

```python
from openai import OpenAI

client = OpenAI()

def text_summarizer(prompt):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are an analytical text assistant. "
                    "Given a passage, extract clear, specific keywords that capture "
                    "key entities, locations, events, cultural references, and time periods. "
                    "Avoid full sentences. Return keywords as a comma-separated list."
                )
            },

            # --- Example 1 ---
            {
                "role": "user",
                "content": (
                    "A flying saucer seen near a guest house, a 7ft alien-like figure emerging "
                    "from a hedge, and a cigar-shaped UFO spotted near a school yard. "
                    "These were among hundreds of reported extraterrestrial encounters "
                    "in a remote Welsh village in the late 1970s."
                )
            },
            {
                "role": "assistant",
                "content": (
                    "flying saucer, alien-like figure, cigar-shaped UFO, school yard, "
                    "extraterrestrial encounters, Welsh village, late 1970s"
                )
            },

            # --- Example 2 ---
            {
                "role": "user",
                "content": (
                    "Each April in a Sri Lankan village, farmers gather near a large reservoir "
                    "to perform rituals seeking blessings for the rice harvest. "
                    "These ancient irrigation systems have supported agriculture for centuries."
                )
            },
            {
                "role": "assistant",
                "content": (
                    "April, Sri Lanka, village ritual, reservoir, rice harvest, "
                    "irrigation systems, agriculture, centuries-old tradition"
                )
            },

            # --- Actual prompt ---
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.5,
        max_tokens=256,
    )

    return response.choices[0].message.content.strip()
```

```python
prompt = (
    "During an early morning kayak trip along a protected coastal lagoon, "
    "our guide paused and motioned toward the shallow water below. "
    "Just beneath the surface, two large sea turtles moved slowly in parallel, "
    "their synchronized movements suggesting a courtship ritual. "
    "Understanding the behavior beforehand transformed what could have been "
    "a brief sighting into a meaningful moment, as we drifted quietly and "
    "observed the interaction unfold in near silence."
)

print(prompt)
```

```python
text_summarizer(prompt)
```

## Poetic Chatbot

```python
from openai import OpenAI

client = OpenAI()

def poetic_chatbot(prompt):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a reflective, lyrical assistant. "
                    "Answer questions using gentle metaphors and flowing language, "
                    "without sounding childish or overly dramatic."
                )
            },

            # --- Example 1 ---
            {
                "role": "user",
                "content": "When was the internet first made widely available?"
            },
            {
                "role": "assistant",
                "content": (
                    "In the quiet years before the millennium turned, "
                    "wires began to hum with shared intent. "
                    "By the early 1990s, the internet stepped beyond laboratories, "
                    "opening its doors to the wider world."
                )
            },

            # --- Example 2 ---
            {
                "role": "user",
                "content": "What causes the changing of the seasons?"
            },
            {
                "role": "assistant",
                "content": (
                    "As the Earth leans and slowly circles the sun, "
                    "light shifts its angle like a wandering gaze. "
                    "From that gentle tilt are born the seasons, "
                    "each arriving in turn, never rushed."
                )
            },

            # --- Actual prompt ---
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=1.0,
        max_tokens=256,
    )

    return response.choices[0].message.content.strip()
```

```python
prompt = (
    "How did cheese first come into existence, "
    "and what moments in early human history led to its creation?"
)

poetic_chatbot(prompt)
```

```python
prompt = (
    "What kind of course do you imagine might be added next to an online "
    "learning platform focused on data science, and why would it matter?"
)

poetic_chatbot(prompt)
```

### Pre-trained Tokenizers

```python
from transformers import AutoTokenizer
```

#### Bert model

```python
model = "bert-base-uncased"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
sentence = "The weather today is amazing, I love sunny days like this"
```

```python
input_ids = tokenizer(sentence)
print(input_ids)
```

```python
# Divided into tokens (text format, easy to read).
tokens = tokenizer.tokenize(sentence)
```

```python
print(tokens)
```

```python
## Convert tokens into numbers (token IDs) according to the model's vocabulary.
token_ids = tokenizer.convert_tokens_to_ids(tokens)
```

```python
print(token_ids)
```

```python
# Create some decoded IDs
# Use tokenizer decode over the token ids

decoded_ids = tokenizer.decode(token_ids)
print(decoded_ids)
```

```python
tokenizer.decode(101)
```

```python
tokenizer.decode(102)
```

## Huggingface and Pytorch/Tensorflow

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
```

```python
print(sentence)
print(input_ids)
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
```

```python
input_ids_pt = tokenizer(sentence, return_tensors = "pt")
print(input_ids_pt)
```

```python
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
```

```python
with torch.no_grad():
    logits = model(**input_ids_pt).logits

predicted_class_id = logits.argmax().item()
model.config.id2label[predicted_class_id]
```
